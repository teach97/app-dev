```python
from pydantic import BaseModel

# 1. 데이터 모델 정의
class UserRegister(BaseModel):
 user_id: int # 필수 입력, 정수형
 username: str # 필수 입력, 문자열
 email: str # 필수 입력, 문자열
 is_admin: bool = False # 선택 입력, 기본값은 False

# 2. 올바른 데이터 입력 및 객체 생성
# 외부에서 입력된 데이터라고 가정합니다 (예: API 요청 데이터)
# 문자열로 들어왔지만 int로 자동 변환 (Parsing)됩니다.
input_data = {
 "user_id": "1004",
 "username": "codemaster",}

# 2. 올바른 데이터 입력 및 객체 생성
# 외부에서 입력된 데이터라고 가정합니다 (예: API 요청 데이터)
# 문자열로 들어왔지만 int로 자동 변환(Parsing)됩니다.
# is_admin은 생략되었으므로 기본값 False가 적용됩니다.
input_data = {"user_id": "1004", "username": "codemaster", "email": "code@example.com"}
user = UserRegister(**input_data)
print("생성된 객체:", user)
print("유저 ID 타입:", type(user.user_id)) # <class 'int'> 출력

# 3. 데이터 출력 및 직렬화 (Serialization)
# DB에 저장하거나 타 시스템으로 전송하기 위해 딕셔너리나 JSON으로 변환합니다.
user_dict = user.model_dump()
user_json = user.model_dump_json()
print("딕셔너리 변환:", user_dict)
print("JSON 문자열 변환:", user_json)
```
